```python
import pandas as pd
import html
from bs4 import BeautifulSoup

def clean_html(text):
    if pd.isna(text):
        return ""

    # Decode HTML entities (&lt; -> <)
    decoded = html.unescape(text)

    # Remove HTML tags
    soup = BeautifulSoup(decoded, "html.parser")

    return soup.get_text(separator=" ", strip=True)


df = pd.read_csv("/Users/shashank/Documents/Projects/ai-job-market-dashboard/data/jobs_master.csv")

sample = df["description"].iloc[0]

print(clean_html(sample)[:1000])
```
